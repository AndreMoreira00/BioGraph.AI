## Features:

- rfam_acc (Rfam Accession Number)
  
  O que é: O número de acesso da família Rfam. É um identificador único e estável para cada família de RNA na base de dados Rfam (ex: RF00162 para a família T-box). 
  
  Importância: Permite que você agrupe todas as estruturas que pertencem à mesma família de RNA. É a sua chave de busca principal para a Rfam.

- pdb_id (PDB Identifier)

  O que é: O identificador único (geralmente de 4 caracteres) de uma entrada na base de dados Protein Data Bank (PDB). Cada pdb_id corresponde a um arquivo que descreve a estrutura 3D de uma ou mais moléculas biológicas.

  Importância: É o seu link para a "verdade experimental". Com este ID, você pode baixar o arquivo .pdb correspondente e visualizar a estrutura 3D real da molécula de RNA.

- chain

  O que é: A letra ou símbolo que identifica uma molécula específica (uma cadeia polimérica) dentro de um arquivo PDB. Um único arquivo PDB pode conter múltiplas cadeias de proteínas ou ácidos nucleicos.

  Importância: É crucial porque diz exatamente qual molécula dentro do arquivo PDB corresponde à família Rfam. Sem isso, você não saberia qual parte da estrutura olhar.

- pdb_start e pdb_end

  O que é: As coordenadas (número do primeiro e do último resíduo/nucleotídeo) que definem a região específica na cadeia do PDB que corresponde à família Rfam.

  Importância: Muitas vezes, a molécula de RNA no PDB é maior do que o domínio funcional descrito pela família Rfam. Estes números delimitam com precisão o trecho exato da estrutura que você deve analisar.

- bit_score

  O que é: Uma pontuação normalizada que mede a qualidade do alinhamento entre a sequência do PDB e o modelo da família Rfam.

  Importância: Quanto maior o bit_score, melhor e mais confiável é a correspondência. É uma medida direta da similaridade. É útil para filtrar e manter apenas os mapeamentos de alta qualidade.

- evalue_score (E-value)

  O que é: O "Valor Esperado" (Expect Value). É uma medida estatística que indica o número de correspondências com um bit_score igual ou superior que você esperaria encontrar por puro acaso ao pesquisar em uma base de dados de tamanho semelhante.

  Importância: Quanto menor o E-value, mais significativa é a correspondência. Um E-value muito baixo (ex: 1.5e-20) significa que é extremamente improvável que a correspondência seja uma coincidência. É o principal indicador de significância estatística.

- cm_start e cm_end

  O que é: As coordenadas de início e fim que descrevem qual parte do "Modelo de Covariância" (CM) da Rfam alinhou com a sequência do PDB. O CM é o modelo computacional que define a família de RNA.

  Importância: Informa se a estrutura do PDB corresponde ao modelo inteiro da família (ex: de 1 a 100 em um modelo de 100 posições) ou apenas a um fragmento dele. Uma correspondência completa (cm_start~1 e cm_end~tamanho do modelo) é geralmente mais valiosa.

- hex_colour

  O que é: Um código de cor hexadecimal (ex: #FF5733).

  Importância: É um dado puramente para visualização, provavelmente usado pelo website da Rfam para colorir diferentes famílias de RNA de forma consistente em gráficos e diagramas. Para a análise de dados e previsão de estruturas, esta coluna pode ser ignorada.
